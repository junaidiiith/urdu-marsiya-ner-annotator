```python
from ner_annotator.utils import get_chunks_response
from ner_annotator.utils import CLAUDE_MODELS, OPENAI_MODELS, calculate_hash, DEEPSEEK_MODELS
from ner_annotator.llm_tagger import TaggedElements, NERMode, get_ner_prompt_messages_per_chunk
import os
from tqdm.auto import tqdm


big_files = [
    "34. AYE-MOMINO-ROO-O-KE-MUSEEBAT-KE-DIN-AAYE",
    "141. JAB-SHAH-KO-MOHLAT-NA-MILI-TOF-E-HARAM-KI",
    "4. AALAM-MAIN-MURTAZA-KI-WILADAT-KI-DHOO0M-HAI",
    "87. JAB-AASMAN-PE-KHATM-HOA-DOR-E-JAM-E-SHAB",
    "38. AYE-TABA-RESA-KHULD-KA-GULZAR-DIKHAAA-DE"
]
```

```python
all_llms = DEEPSEEK_MODELS + OPENAI_MODELS
modes = [v for v in NERMode]
for llm in all_llms:
    if llm.startswith("o"):
        continue
    for big_file in tqdm(big_files):
        with open(f"dataset/marsiya-all/{big_file}.pdf.json.txt", 'r', encoding='utf-8') as file:
            text = file.read()
        print(f"Processing {big_file} with {llm}")
        for mode in modes:
            chunked_messages = get_ner_prompt_messages_per_chunk(text, mode=mode)[:5]
            ner_tags = get_chunks_response(
                chunked_messages,
                model=llm,
                response_format=TaggedElements,
                tqdm=tqdm
            )
            break
        break
    break
```
